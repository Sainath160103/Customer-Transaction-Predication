```python
#Separate Features and Target
X = train.drop("target", axis=1)

y = train["target"]

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python
# Split Dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)

print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed!")
```

```python
#Logistic Regression
lr = LogisticRegression(
    random_state=42,
    max_iter=1000
)

lr.fit(
    X_train_scaled,
    y_train
)

lr_pred = lr.predict(X_test_scaled)

print("Logistic Regression training completed!")
```

```python
# Decision Tree
dt = DecisionTreeClassifier(
    random_state=42
)

dt.fit(
    X_train,
    y_train
)

dt_pred = dt.predict(X_test)

print("Decision Tree training completed!")
```

```python
#Random Forest
rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf.fit(
    X_train,
    y_train
)

rf_pred = rf.predict(X_test)

print("Random Forest training completed!")
```

```python
#KNN training
knn = KNeighborsClassifier(
    n_neighbors=5,
    n_jobs=-1
)

knn.fit(
    X_train_scaled,
    y_train
)

knn_pred = knn.predict(X_test_scaled)

print("KNN training completed!")
```

```python
# Use a smaller sample for SVM to reduce training time

svm_sample_size = 20000

X_svm, _, y_svm, _ = train_test_split(
    X_train_scaled,
    y_train,
    train_size=svm_sample_size,
    random_state=42,
    stratify=y_train
)

print("SVM Training Data:", X_svm.shape)
```

```python
#SVM Training Data
svm = SVC(
    kernel="rbf",
    probability=True,
    random_state=42
)

svm.fit(
    X_svm,
    y_svm
)

svm_pred = svm.predict(X_test_scaled)

print("SVM training completed!")
```

```python
#XGBoost
xgb = XGBClassifier(
    n_estimators=100,
    max_depth=6,
    learning_rate=0.1,
    random_state=42,
    eval_metric="logloss",
    n_jobs=-1
)

xgb.fit(
    X_train,
    y_train
)

xgb_pred = xgb.predict(X_test)

print("XGBoost training completed!")
```

```python
#Model Evaluation
#Create a function to calculate metrics.
def evaluate_model(name, y_true, y_pred, y_prob=None):

    accuracy = accuracy_score(
        y_true,
        y_pred
    )

    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )

    if y_prob is not None:
        auc = roc_auc_score(
            y_true,
            y_prob
        )
    else:
        auc = np.nan

    return {
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1,
        "ROC-AUC": auc
    }
```

```python
#Generate Model Results
results = []

# Logistic Regression
results.append(
    evaluate_model(
        "Logistic Regression",
        y_test,
        lr_pred,
        lr.predict_proba(X_test_scaled)[:, 1]
    )
)

# Decision Tree
results.append(
    evaluate_model(
        "Decision Tree",
        y_test,
        dt_pred,
        dt.predict_proba(X_test)[:, 1]
    )
)

# Random Forest
results.append(
    evaluate_model(
        "Random Forest",
        y_test,
        rf_pred,
        rf.predict_proba(X_test)[:, 1]
    )
)

# KNN
results.append(
    evaluate_model(
        "KNN",
        y_test,
        knn_pred,
        knn.predict_proba(X_test_scaled)[:, 1]
    )
)

# SVM
results.append(
    evaluate_model(
        "SVM",
        y_test,
        svm_pred,
        svm.predict_proba(X_test_scaled)[:, 1]
    )
)

# XGBoost
results.append(
    evaluate_model(
        "XGBoost",
        y_test,
        xgb_pred,
        xgb.predict_proba(X_test)[:, 1]
    )
)

comparison = pd.DataFrame(results)

comparison
```

```python
#Sort Models by F1-Score
comparison = comparison.sort_values(
    by="F1-Score",
    ascending=False
).reset_index(drop=True)

comparison
```

```python
#Model Accuracy Chart
plt.figure(figsize=(10,6))

plt.bar(
    comparison["Model"],
    comparison["Accuracy"]
)

plt.title("Model Accuracy Comparison")

plt.xlabel("Model")

plt.ylabel("Accuracy")

plt.xticks(
    rotation=30,
    ha="right"
)

plt.tight_layout()

plt.show()
```

```python
#Model Accuracy Chart
plt.figure(figsize=(10,6))

plt.bar(
    comparison["Model"],
    comparison["F1-Score"]
)

plt.title("Model F1-Score Comparison")

plt.xlabel("Model")

plt.ylabel("F1-Score")

plt.xticks(
    rotation=30,
    ha="right"
)

plt.tight_layout()

plt.show()
```

```python
#XGBoost Confusion Matrix
cm_xgb = confusion_matrix(
    y_test,
    xgb_pred
)

plt.figure(figsize=(6,5))

sns.heatmap(
    cm_xgb,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title("XGBoost Confusion Matrix")

plt.xlabel("Predicted")

plt.ylabel("Actual")

plt.show()
```

```python
#Random Forest Confusion Matrix
cm_rf = confusion_matrix(
    y_test,
    rf_pred
)

plt.figure(figsize=(6,5))

sns.heatmap(
    cm_rf,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title("Random Forest Confusion Matrix")

plt.xlabel("Predicted")

plt.ylabel("Actual")

plt.show()
```

```python
#Classification Report
print("XGBoost Classification Report")

print(
    classification_report(
        y_test,
        xgb_pred,
        zero_division=0
    )
)
```

```python
#ROC Curve
models_roc = {
    "Logistic Regression": (
        y_test,
        lr.predict_proba(X_test_scaled)[:, 1]
    ),

    "Random Forest": (
        y_test,
        rf.predict_proba(X_test)[:, 1]
    ),

    "XGBoost": (
        y_test,
        xgb.predict_proba(X_test)[:, 1]
    )
}

plt.figure(figsize=(8,6))

for name, (y_true, y_prob) in models_roc.items():

    fpr, tpr, _ = roc_curve(
        y_true,
        y_prob
    )

    auc_score = roc_auc_score(
        y_true,
        y_prob
    )

    plt.plot(
        fpr,
        tpr,
        label=f"{name} (AUC={auc_score:.3f})"
    )

plt.plot(
    [0,1],
    [0,1],
    linestyle="--"
)

plt.title("ROC Curve Comparison")

plt.xlabel("False Positive Rate")

plt.ylabel("True Positive Rate")

plt.legend()

plt.show()
```

```python
#Select Best Model
#We'll select the best model based on F1-score.
best_model_name = comparison.iloc[0]["Model"]

print("Best Model:", best_model_name)
```

```python
#Test Saved Model
#Because our models use different preprocessing:
#Logistic Regression, KNN, and SVM use scaled data.
#Decision Tree, Random Forest, and XGBoost use unscaled data.
if best_model_name in [
    "Logistic Regression",
    "KNN",
    "SVM"
]:

    final_X_test = loaded_scaler.transform(X_test)

else:

    final_X_test = X_test

# Make predictions
final_predictions = loaded_model.predict(
    final_X_test
)

# Accuracy
final_accuracy = accuracy_score(
    y_test,
    final_predictions
)

print(
    "Final Saved Model Accuracy:",
    round(final_accuracy * 100, 2),
    "%"
)
```

```python
#Final Classification Report
print(
    classification_report(
        y_test,
        final_predictions,
        zero_division=0
    )
)
```

```python
#Final Confusion Matrix
final_cm = confusion_matrix(
    y_test,
    final_predictions
)

plt.figure(figsize=(6,5))

sns.heatmap(
    final_cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title(
    f"{best_model_name} - Final Confusion Matrix"
)

plt.xlabel("Predicted")

plt.ylabel("Actual")

plt.show()
```
